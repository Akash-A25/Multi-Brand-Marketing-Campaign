```python
import pandas as pd
```

```python
# read filles in pandas dataset1
df = pd.read_csv('nykaa_campaign_data_with_nulls.csv')
```

```python
# read filles in pandas dataset3
df2 = pd.read_csv('tira_campaign_data_with_nulls.csv')
```

```python
# there data set combined single data set
df_comb = pd.concat([df,df1,df2],ignore_index= True)
```

```python
df_comb
```

```python
#drop the Date and Campaign_ID
df_comb.drop(columns=['Campaign_ID','Date'], inplace = True)
```

```python
# cheak data types
df_comb.info()
```

```python
# cheaking null value 
df_comb.isnull().sum()
```

```python
# drop the text null data row 
df_comb.dropna(subset=['Campaign_Type','Target_Audience','Channel_Used','Language','Customer_Segment'],inplace=True)
```

```python
#index number re-assing
df_comb.reset_index(drop=True,inplace=True)
```

```python
# bring the company name as the first colums
cols = ['Company_Name'] + [ col for col in df_comb.columns if col != 'Company_Name']
df_comb = df_comb[cols]
```

```python
df_comb.head()
```

```python
df_comb.info()
```

```python
df_comb.isnull().sum()
```

```python
# find Duration mean_value using groupby 
Duration_mean = df_comb.groupby(['Company_Name','Campaign_Type','Target_Audience'])['Duration'].mean()
print(Duration_mean)
```

```python
#Duration_mean_value transfrom flot value
Duration_mean_value = float(df_comb['Duration'].mean())
```

```python
# Duration fill null value
df_comb['Duration'] = df_comb['Duration'].fillna(Duration_mean_value)
```

```python
df_comb.isnull().sum()
```

```python
# find Impressions mean_value using groupby 
Impressions_mean = df_comb.groupby(['Company_Name','Channel_Used','Duration'])['Impressions'].mean()
print(Impressions_mean)
```

```python
# Impressions_mean_value transfrom float
Impressions_mean_value = float(df_comb['Impressions'].mean())
```

```python
# filling nall value 
df_comb['Impressions'] = df_comb['Impressions'].fillna(Impressions_mean_value)
```

```python
df_comb.isnull().sum()
```

```python
# find Clicks mean_value using groupby 
Clicks_mean = df_comb.groupby(['Company_Name','Channel_Used','Target_Audience'])['Clicks'].mean()
print(Clicks_mean)
```

```python
# cliks_mean_value transfrom float
Cliks_mean_value = float(df_comb['Clicks'].mean())
```

```python
df_comb['Clicks'] = df_comb['Clicks'].fillna(Cliks_mean_value)
```

```python
df_comb.isnull().sum()
```

```python
# finding clicks to leads %
global_ratio = df_comb['Leads'].sum()/df_comb['Clicks'].sum()

print(global_ratio)
```

```python
# takes lead global_ratio from lead and multiples it by the clicks
df_comb['Leads'] = df_comb['Leads'].fillna(df_comb['Clicks']*global_ratio)
```

```python
df_comb.head()
```

```python
df_comb.isnull().sum()
```

```python
# Conversions rate finding
Conversions_rate = df_comb.groupby(['Company_Name','Channel_Used'])['Conversions'].transform('sum')/df_comb.groupby(['Company_Name','Channel_Used'])['Leads'].transform('sum')
print(Conversions_rate)
```

```python
df_comb['Conversions']=df_comb['Conversions'].fillna(df_comb['Leads']*Conversions_rate)
```

```python
df_comb.isnull().sum()
```

```python
df_comb['Revenue'] = df_comb['Revenue'].fillna(0)
```

```python
df_comb.isnull().sum()
```

```python
cost_to_revenue_ratio = df_comb.groupby(['Company_Name','Channel_Used'])['Acquisition_Cost'].transform('mean')
print(cost_to_revenue_ratio)
```

```python
df_comb['Acquisition_Cost'] = df_comb['Acquisition_Cost'].fillna( cost_to_revenue_ratio)
```

```python
df_comb.isnull().sum()
```

```python
# roi drop it not correct value
df_comb.drop(columns=['ROI'],inplace=True)
```

```python
df_comb['ROI'] =''
```

```python
df_comb['ROI']= ((df_comb['Revenue']-df_comb['Acquisition_Cost'])/df_comb['Acquisition_Cost'])
```

```python
df_comb.isnull().sum()
```

```python
df_comb['Engagement_Score'] = df_comb['Engagement_Score'].fillna(df_comb.groupby(['Company_Name','Campaign_Type','Target_Audience']
                                                                                 )['Engagement_Score'].transform('mean'))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
x=df3.drop(columns=['Revenue'])
y=df3['Revenue']
```

```python
# categorical Columns one-hot encoding
categorical_cols= x.select_dtypes(include=['object','string','category']).columns.tolist()

preprocessor =ColumnTransformer(transformers=[('cat',OneHotEncoder(handle_unknown='ignore',sparse_output=False)
                                              ,categorical_cols)],remainder='passthrough'
                                              )
```

```python
# train \test split
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,mean_absolute_percentage_error,root_mean_squared_error

def evaluate(name,y_test,y_pred):
    r2 = r2_score(y_test,y_pred)
    mae = mean_absolute_error(y_test,y_pred)
    mse = mean_squared_error(y_test,y_pred)
    mape = mean_absolute_percentage_error(y_test,y_pred)
    rmse = root_mean_squared_error(y_test,y_pred)
    return {'Model':name,'R2':r2,'MAE':mae,'MSE':mse,'MAPE':mape,'RMSE':rmse}
```

```python
from sklearn.compose import TransformedTargetRegressor

models = {
    "LinearRegression": LinearRegression(),
    "DecisionTreeRegressor": DecisionTreeRegressor(random_state=20),
    "RandomForestRegressor": RandomForestRegressor(n_estimators=50, random_state=40)
}

results = []
```

```python
from sklearn.pipeline import Pipeline

results = []

best_pipeline = None
best_model_name = None
best_r2 = float('-inf')

for name, model in models.items():
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('model', model)
    ])
    pipeline.fit(x_train, y_train)

    y_pred = pipeline.predict(x_test)
    res = evaluate(name, y_test, y_pred)
    results.append(res)

    if res['R2'] > best_r2:
        best_r2 = res['R2']
        best_pipeline = pipeline
        best_model_name = name

results_df = pd.DataFrame(results).sort_values(by='R2', ascending=False)
display(results_df)

print(f'Best model: {best_model_name} with R2 = {best_r2:.4f}')
```

```python
feature_names = best_pipeline.named_steps['preprocessor'].get_feature_names_out()
print(feature_names)
```
